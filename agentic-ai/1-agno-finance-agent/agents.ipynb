## Import Libraries

```python
from agno.agent import Agent
from agno.models.openai import OpenAIChat
from agno.tools.duckduckgo import DuckDuckGoTools
from agno.tools.yfinance import YFinanceTools
from dotenv import load_dotenv

load_dotenv()
```

## First Agent

```python
web_agent = Agent(
    name="Web Agent",
    model=OpenAIChat(id="gpt-4o"),
    tools=[DuckDuckGoTools()],
    instructions=["Always include sources"],
    markdown=True,
)

result = web_agent.run("what is yfinance. give a short sentence")
print(result.content)
```

## Second Agent

```python
finance_agent = Agent(
    name="Finance Agent",
    role="Get financial data",
    model=OpenAIChat(id="gpt-4o"),
    tools=[YFinanceTools()],
    instructions=["Use tables to display data"],
    markdown=True
)

result = finance_agent.run("what is last day of NVIDIA stock price that your memory has?")
print(result.content)
```

## Team of agents

```python
agent_team = Agent(
    name="Agent Team",
    model=OpenAIChat(id="gpt-4o"),
    stream=[web_agent, finance_agent],
    instructions=["Always include sources", "Use tables to display data"],
    markdown=True
)

for chunk in agent_team.run("Summarize analyst recommendations and share the latest news for NVIDIA", stream=True):
    print(chunk.content, end="", flush=True)
```
